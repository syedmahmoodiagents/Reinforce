# One more Way

```python
env = gym.make("FrozenLake-v1", is_slippery=False)  # deterministic for clarity

# Initialize value function
V = np.zeros(env.observation_space.n)
gamma = 0.99
theta = 1e-6  # convergence threshold

while True:
    delta = 0
    for s in range(env.observation_space.n):
        v = V[s]
        action_values = []

        # For each action, calculate expected return
        for a in range(env.action_space.n):
            q_sa = 0
            for prob, next_state, reward, done in env.P[s][a]:
                q_sa += prob * (reward + gamma * V[next_state])
            action_values.append(q_sa)

        # Bellman optimality update
        V[s] = max(action_values)

        delta = max(delta, abs(v - V[s]))

    if delta < theta:
        break

print("Optimal Value Function:")
print(V)

# Derive optimal policy from value function
policy = np.zeros(env.observation_space.n, dtype=int)
for s in range(env.observation_space.n):
    action_values = []
    for a in range(env.action_space.n):
        q_sa = sum([prob * (reward + gamma * V[next_state])
                    for prob, next_state, reward, done in env.P[s][a]])
        action_values.append(q_sa)
    policy[s] = np.argmax(action_values)

print("\nOptimal Policy:")
print(policy)
```
